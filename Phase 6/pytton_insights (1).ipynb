```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

file_path = r"C:\Users\Admin\Downloads\all cleaned after python.xlsx"
df_eng = pd.read_excel(file_path, sheet_name='engagement')
df_act = pd.read_excel(file_path, sheet_name='activity')
df_child = pd.read_excel(file_path, sheet_name='child')
df_life = pd.read_excel(file_path, sheet_name='life_style')

eng_act = pd.merge(df_eng, df_act, on='activity_id', how='inner')
eng_child = pd.merge(df_eng, df_child, on='child_id', how='inner')
eng_life = pd.merge(df_eng, df_life, on='child_id', how='inner')
```

```python
# 1. Duration vs Engagement
df_act['duration_category'] = pd.qcut(
    df_act['duration'], 
    q=3, 
    labels=['Shorter Duration', 'Medium Duration', 'Longer Duration']
)
insight_1 = pd.merge(
    df_eng[['session_id', 'child_id', 'activity_id', 'engagement_score c']], 
    df_act[['activity_id', 'duration', 'duration_category']], 
    on='activity_id', how='left'
)

# 2. Group Size vs Engagement
df_act['group_size_category'] = pd.cut(
    df_act['group_size'], 
    bins=[-1, 1, 5, 15, 100], 
    labels=['Individual (1)', 'Small Group (2-5)', 'Medium Group (6-15)', 'Large Group (>15)']
)
insight_2 = pd.merge(
    df_eng[['session_id', 'child_id', 'activity_id', 'engagement_score c']], 
    df_act[['activity_id', 'group_size', 'group_size_category']], 
    on='activity_id', how='left'
)

# 3. Movement Required vs Engagement
df_act['movement_label'] = df_act['movement_required'].map({0: 'No Movement', 1: 'Movement Required'})
insight_3 = pd.merge(
    df_eng[['session_id', 'child_id', 'activity_id', 'engagement_score c']], 
    df_act[['activity_id', 'movement_required', 'movement_label']], 
    on='activity_id', how='left'
)

# 4. Age vs Engagement
df_child['age_group'] = pd.cut(
    df_child['age c'], 
    bins=[0, 6, 11, 18], 
    labels=['Young Children (4-6)', 'Middle Children (7-11)', 'Teens (12-18)']
)
insight_4 = pd.merge(
    df_eng[['session_id', 'child_id', 'engagement_score c']], 
    df_child[['child_id', 'age c', 'age_group']], 
    on='child_id', how='left'
)

# 5. School Type vs Engagement
insight_5 = pd.merge(
    df_eng[['session_id', 'child_id', 'engagement_score c']], 
    df_child[['child_id', 'school type c']], 
    on='child_id', how='left'
)

# 6. Environment Type vs Engagement
insight_6 = pd.merge(
    df_eng[['session_id', 'child_id', 'engagement_score c']], 
    df_child[['child_id', 'environment_type c']], 
    on='child_id', how='left'
)

# 7. Household Income vs Engagement
df_child['income_tier'] = pd.qcut(
    df_child['household_income'], 
    q=3, 
    labels=['Low Income', 'Medium Income', 'High Income']
)
insight_7 = pd.merge(
    df_eng[['session_id', 'child_id', 'engagement_score c']], 
    df_child[['child_id', 'household_income', 'income_tier']], 
    on='child_id', how='left'
)

# 8. Technology Access vs Engagement
insight_8 = pd.merge(
    df_eng[['session_id', 'child_id', 'engagement_score c']], 
    df_child[['child_id', 'access_to_technology C']], 
    on='child_id', how='left'
)

# 9. Number of Devices vs Engagement
insight_9 = pd.merge(
    df_eng[['session_id', 'child_id', 'engagement_score c']], 
    df_child[['child_id', 'number_of_devices']], 
    on='child_id', how='left'
)

# 10. Lifestyle Factors vs Engagement
lifestyle_cols = ['sleep_duration', 'screen_time', 'gaming', 'social_media_usage', 'outdoor_time', 'study_time']
insight_10 = pd.merge(
    df_eng[['session_id', 'child_id', 'engagement_score c']], 
    df_life[['child_id'] + lifestyle_cols], 
    on='child_id', how='left'
)
```
