```python
print(df9.isnull().sum())

print("\n إجمالي الخلايا المفقودة:", df9.isnull().sum().sum())

df9[df9.isnull().any(axis=1)]
```

```python


# 1- parents_working_hours (43 فاضي) -> متوسط حسب حجم العيلة
df9['parents_working_hours'] = df9.groupby('family_size')['parents_working_hours'].transform(
    lambda x: x.fillna(x.median())
)
# لو لسه فاضي لان العيلة كلها فاضية
df9['parents_working_hours'] = df9['parents_working_hours'].fillna(df9['parents_working_hours'].median())

# 2- household_income (89 فاضي) -> الوسيط حسب البيئة / نوع المدرسة
df9['household_income'] = df9.groupby('environment_type c')['household_income'].transform(
    lambda x: x.fillna(x.median())
)
# لو لسه فاضي
df9['household_income'] = df9['household_income'].fillna(df9['household_income'].median())

# 3- تأكيد ان كله اتصلح
print("=== بعد الحل ===")
print(df9[['parents_working_hours','household_income']].isna().sum())
print(f"إجمالي الخلايا المفقودة: {df9.isna().sum().sum()}") # لازم 0

# 4- شوف لو فيه مشاكل تانية
print(f"\nعدد المكرر: {df9.duplicated().sum()}")
```

'activity'

```python

df10 = temp['activity']
df10.head()
```

```python
print(df10.isnull().sum())

print("\n إجمالي الخلايا المفقودة:", df10.isnull().sum().sum())

df10[df10.isnull().any(axis=1)]
```

```python

for col in ['interaction_type', 'activity_format', 'participation_type']:
    most_common = df10[col].mode()[0] # بيجيب اكتر قيمة متكررة
    df10[col] = df10[col].fillna(most_common)

# 2- difficulty_level c (32 فاضي) -> ده مرتبط بنوع النشاط
# صعوبة النشاط حسب نوع النشاط نفسه
df10['difficulty_level c'] = df10.groupby('activity type')['difficulty_level c'].transform(
    lambda x: x.fillna(x.mode()[0] if not x.mode().empty else x.mode())
)

# لو لسه فيه فاضي بعد التجميع
df10['difficulty_level c'] = df10['difficulty_level c'].fillna(df10['difficulty_level c'].mode()[0])

# 3- تأكد
print(df10[['interaction_type','activity_format','participation_type','difficulty_level c']].isna().sum())
print(f"إجمالي المفقود: {df10.isna().sum().sum()}") 
```
